## Separate keeper domains

Per-90 values are 90 × recorded totals / reported keeper minutes. Cross stopping uses stopped/faced counts, with nonpositive or invalid denominators unavailable. PSxG/90 describes shot difficulty faced, not ability by itself. Saves and sweeping depend on opportunities and team tactics. Pass-attempt labels differ between source archives. Plots separate season/source groups; PSxG fills receive their own source group. No composite score, cross-season leaderboard or replacement value is produced.

```python
chart = plot_domains(frame)
display(Image(filename=str(chart)))
display(distributions.head(24))
```
