Core question: observed seasonal contribution above a readily available replacement-level player. This first step examines measurement coverage and overlap before any contribution formula is chosen. Different positions need different evidence; eventual comparison requires a validated common outcome scale. No score, weights, replacement baseline or league adjustment is selected here.

## Candidate contribution domains

- Attacking: goals, non-penalty goals, assists, npxG, xA and xAG.
- Progression: progressive passes and carries.
- Defensive actions: tackles and interceptions; these omit much off-ball prevention.
- Keeper actions: reported shot-stopping residual, distribution, cross stopping and sweeping. PSxG faced and shot counts also describe opportunity/difficulty rather than ability alone.

These are measurement candidates, not independently additive contributions. More events do not automatically mean greater value. Goals will not be downweighted merely to force a balanced leaderboard.

## Metric overlap within context

Spearman correlations are descriptive checks within position × league × season × source. Pair counts and constant/insufficient samples are shown. Three pairs is only a computability guard, not evidence of reliability. Older-source PSxG fills are omitted only from these source-cohort correlations; they remain in descriptive profiles. Source versions remain unconfirmed. Correlations cannot determine importance, weights, causality or league strength.
